# 1 - Containerization and Infastructure as Code


## Docker Commands
- `docker run image-name` runs the container.
- `docker run -it image-name` runs the container and can be interacted with.
  - `-it`: `-i`nteractive `-t`erminal
  - Provides an *interactive terminal* inside the container.

You can view the current running containers in **Docker.Desktop**.

`docker run -it python:3.13.11-slim`
`docker run -it [image-name]:[tag]`

- `rm -rf`: `r`e`m`ove `-r`ecursively `f`orce
  - Never run on host machine.
  - Run only run in containers if a mistake is made.

- `docker ps -aq`: List containers `-a`ll containers `q`uiet mode

`docker run -it --entrypoint=bash -v $(pwd)/test:/app/test python:3.13.11-slim`
- Starts an interactive Python container with a bash shell and mounts a folder(s) called `/app/test`
- `--entrypoint=bash`: Use bash instead of default Python
- `-v $(pwd)/test:/app/test`:mount local test folder to `/app/test` in container
- `python:3.13.11-slim`: The image
